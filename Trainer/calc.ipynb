<h1> Trainer Dataset Fugazi Mean </h1>

```python
"""
Memória
"""
MEM = [12.08, 12.05, 12.05]

MEAN = np.mean (MEM)
DP = np.std (MEM)
print (f"{MEAN:.2F} ± {DP:.4f}")

"""
Tempo
Aqui usamos Tempo por Effective Batch (Step) * (Número de Exemplos / Gradient Accumulation)
Desta maneira encontramos o tempo para realizar uma Epoch com o Dataset Original. Realizamos este cálculo para as 3 runs. Vamos reportar os valores em horas.
"""
STEPS_EPOCH = 48403 #Número de Steps necessários para realizar uma epoch. (Número de Exemplos / Gradient Accumulation)
TIME = [66.55, 69.25, 69.24]
TIME_EPOCH = [3221219.65, 3351907.75, 3351423.72] #Segundos
TIME_EPOCH_HORAS = [895, 931, 931]

MEAN = np.mean (TIME_EPOCH_HORAS)
DP = np.std (TIME_EPOCH_HORAS)
print (f"{MEAN:.2F} ± {DP:.4f}")

"""
Throughput
"""
THROUGHPUT = [193.01, 185.53, 186.05]

MEAN = np.mean (THROUGHPUT)
DP = np.std (THROUGHPUT)
print (f"{MEAN:.2F} ± {DP:.4f}")
```
